```sql
SELECT
  count_if(is_flagged) AS flagged,
  min(CASE WHEN is_flagged THEN fraud_score END) AS min_score_flagged,
  max(CASE WHEN NOT is_flagged THEN fraud_score END) AS max_score_not_flagged
FROM ml.fraud_scores_dev
```

```sql
ALTER TABLE ml.fraud_scores_dev
SET TBLPROPERTIES (delta.enableChangeDataFeed = true)
```

```sql
UPDATE ml.fraud_scores_dev
SET is_flagged = true
WHERE fraud_score >= 0.4 AND NOT is_flagged
```

```sql
SELECT _change_type, _commit_version, transaction_id, fraud_score, is_flagged
FROM table_changes('ml.fraud_scores_dev', 2)
ORDER BY transaction_id, _change_type
```
