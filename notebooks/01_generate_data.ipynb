```python
 %sql
SELECT
  COUNT(*)                                        AS total_rows,
  COUNT(DISTINCT transaction_id)                  AS unique_ids,
  SUM(CASE WHEN is_fraud THEN 1 ELSE 0 END)       AS fraud_rows,
  SUM(CASE WHEN amount IS NULL THEN 1 ELSE 0 END) AS null_amounts,
  SUM(CASE WHEN amount < 0 THEN 1 ELSE 0 END)     AS negative_amounts,
  SUM(CASE WHEN currency <> 'MXN' THEN 1 ELSE 0 END) AS bad_currency
FROM read_files('/Volumes/fraud_lakehouse/raw/landing/transactions', format => 'json')
```
